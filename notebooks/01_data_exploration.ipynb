# Data Exploration: Ethiopia Financial Inclusion

This notebook explores the starter dataset `ethiopia_fi_unified_data.csv` to understand its structure, coverage, and quality.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# specific formatting
pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')
```

## 1. Load Data
Loading the unified dataset.

```python
DATA_PATH = '../data/raw/ethiopia_fi_unified_data.csv'

try:
    df = pd.read_csv(DATA_PATH)
    print("Dataset loaded successfully.")
    print(f"Shape: {df.shape}")
except FileNotFoundError:
    print(f"Error: File not found at {DATA_PATH}")
```

## 2. Structure & Schema Checks

```python
if 'df' in locals():
    display(df.head())
    display(df.info())
```

## 3. Distribution by Record Type

```python
if 'df' in locals():
    print("Distribution by Record Type:")
    print(df['record_type'].value_counts())
```

## 4. Observations Analysis
Focusing on rows where `record_type` is 'observation'.

```python
if 'df' in locals():
    observations = df[df['record_type'] == 'observation']
    print(f"\nNumber of observations: {len(observations)}")
    
    print("\nUnique Indicators:")
    print(observations['indicator_code'].unique())
    
    print("\nDate Range:")
    print(f"Start: {observations['observation_date'].min()}")
    print(f"End: {observations['observation_date'].max()}")
```
